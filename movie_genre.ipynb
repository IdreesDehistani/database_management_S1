```python
import pandas as pd
```

```python
junction = pd.read_csv("movie_lookup.csv")
genre = pd.read_csv("genre.csv")

genre_data = junction[['movieid', 'genre']].dropna().copy()

genre_data['genre'] = genre_data['genre'].str.split(',')
print(genre_data)
```

```python
genre_data = genre_data.explode('genre')
print(genre_data)
```

```python
movie_genre = genre_data.merge(
    genre,
    on=['genre'],
    how='left'
)
```

```python
movie_genre.info()
movie_genre.isnull().sum().sum()

movie_genre = movie_genre[['movieid','genreid']]
movie_genre.to_csv("movie_genre.csv",index=False)
```
