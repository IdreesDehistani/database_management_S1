```python
import pandas as pd
```

```python
junction = pd.read_csv("movie_lookup.csv")
awards = pd.read_csv("awards.csv")

awards_data = junction[['movieid', 'awards']].dropna().copy()

awards_data[['Rank', 'Category', 'Year']] = awards_data['awards'].str.extract(
    r'#(\d+)(.*?)of(\d{4})'
)
```

```python
awards.info()
awards_data.info()
```

```python
awards_data['Year'] = awards_data['Year'].astype(int)
```

```python
awards_data['Category'] = (
    awards_data['Category']
    .str.replace(r'(?<=[a-z])(?=[A-Z])', ' ', regex=True)
    .str.strip()
)
```

```python
movie_awards = awards_data.merge(
    awards,
    on=['Category', 'Year'],
    how='left'
)
```

```python
movie_awards.info()
movie_awards.isnull().sum().sum()
movie_awards = movie_awards[['movieid','awardsid','Rank']]
movie_awards.to_csv('movie_awards.csv',index=False)
```
