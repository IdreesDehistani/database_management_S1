```python
# Import all the required libraries
import pandas as pd
```

```python
# Load the Excel File
meta = pd.read_excel("metaClean43Brightspace.xlsx")
```

```python
print(meta.columns)
```

```python
# Remove unused columns
meta = meta.drop(columns=['url','studio','rating','runtime','cast','director','summary','awards','metascore','userscore','RelDate'])
```

```python
meta.info()
```

```python
# meta.to_excel('genre.xlsx', index=False)
```

```python
meta = meta.astype({
    'title': 'string',
    'genre': 'string'
})
```

```python
# Split each genre by comma, flatten the results, and remove duplicates
unique_genres = (
    meta['genre']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .unique()
)

# Sort alphabetically
unique_genres = sorted(unique_genres)

print(unique_genres)
```

```python
genre = pd.DataFrame({'genre':unique_genres})
print(genre)
```

```python
genre.insert(0, 'Genre_ID', range(1,len(genre)+1))
```

```python
print(genre)
```

```python
genre.to_csv('genrefinish.csv', index=False)
```
